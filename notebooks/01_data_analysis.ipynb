```python
# Check the number of unique items, stores, departments and categories

print("Unique items:", sales["item_id"].nunique())
print("Unique departments:", sales["dept_id"].nunique())
print("Unique categories:", sales["cat_id"].nunique())
print("Unique stores:", sales["store_id"].nunique())
print("Unique states:", sales["state_id"].nunique())
```

```python
# Total sales across all days

day_columns = [col for col in sales.columns if col.startswith("d_")]

total_sales = sales[day_columns].sum().sum()

print("Total units sold:", total_sales)
```

```python
# Total sales for each item

sales["total_sales"] = sales[day_columns].sum(axis=1)

top_items = sales[["item_id", "store_id", "total_sales"]].sort_values(
    by="total_sales",
    ascending=False
)

print(top_items.head(10))
```

```python
# Average sales over the last 30 days

recent_days = day_columns[-30:]

sales["avg_last_30_days"] = sales[recent_days].mean(axis=1)

recent_demand = sales[
    ["item_id", "store_id", "avg_last_30_days"]
].sort_values(
    by="avg_last_30_days",
    ascending=False
)

print(recent_demand.head(10))
```

```python
# Calculate expected demand for the next 7 days

sales["expected_7_day_demand"] = (
    sales["avg_last_30_days"] * 7
)

# Recommended replenishment quantity
# Add 20% safety stock

sales["reorder_quantity"] = (
    sales["expected_7_day_demand"] * 1.20
).round().astype(int)

reorder_list = sales[
    ["item_id", "store_id", "avg_last_30_days",
     "expected_7_day_demand", "reorder_quantity"]
].sort_values(
    by="reorder_quantity",
    ascending=False
)

print(reorder_list.head(10))
```

```python
# Create replenishment priority

def priority(avg_demand):
    if avg_demand >= 50:
        return "High"
    elif avg_demand >= 20:
        return "Medium"
    else:
        return "Low"

sales["priority"] = sales["avg_last_30_days"].apply(priority)

priority_summary = sales["priority"].value_counts()

print(priority_summary)
```

```python
# Show the highest-priority products

high_priority = sales[
    sales["priority"] == "High"
][
    ["item_id", "store_id", "avg_last_30_days",
     "expected_7_day_demand", "reorder_quantity", "priority"]
].sort_values(
    by="reorder_quantity",
    ascending=False
)

print(high_priority.head(20))
```

```python
# Calculate demand variability over the last 30 days

sales["daily_std"] = sales[recent_days].std(axis=1)

# Safety stock = 1.65 × standard deviation × sqrt(lead time)
# Assuming a 7-day lead time

sales["safety_stock"] = (
    1.65 * sales["daily_std"] * (7 ** 0.5)
).round()

# Recalculate reorder quantity

sales["reorder_quantity_ai"] = (
    sales["expected_7_day_demand"] + sales["safety_stock"]
).round().astype(int)

ai_reorder_list = sales[
    ["item_id", "store_id", "avg_last_30_days",
     "daily_std", "safety_stock", "reorder_quantity_ai"]
].sort_values(
    by="reorder_quantity_ai",
    ascending=False
)

print(ai_reorder_list.head(15))
```

```python
# Save replenishment recommendations

output_columns = [
    "item_id",
    "store_id",
    "avg_last_30_days",
    "daily_std",
    "safety_stock",
    "expected_7_day_demand",
    "reorder_quantity_ai",
    "priority"
]

replenishment_report = sales[output_columns].copy()

replenishment_report.to_csv(
    "../data/replenishment_recommendations.csv",
    index=False
)

print("Replenishment report saved successfully!")
print("Rows:", len(replenishment_report))
```

```python
# Replenishment summary by store

store_summary = replenishment_report.groupby("store_id").agg(
    total_reorder_quantity=("reorder_quantity_ai", "sum"),
    average_daily_demand=("avg_last_30_days", "mean"),
    total_safety_stock=("safety_stock", "sum")
).sort_values(
    by="total_reorder_quantity",
    ascending=False
)

print(store_summary)
```

```python
# Replenishment summary by priority

priority_summary = replenishment_report.groupby("priority").agg(
    products=("item_id", "count"),
    total_reorder_quantity=("reorder_quantity_ai", "sum")
).sort_values(
    by="total_reorder_quantity",
    ascending=False
)

print(priority_summary)
```

```python
import matplotlib.pyplot as plt

priority_counts = replenishment_report["priority"].value_counts()

plt.figure(figsize=(7, 5))
priority_counts.plot(kind="bar")

plt.title("Products by Replenishment Priority")
plt.xlabel("Priority")
plt.ylabel("Number of Products")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
top_10 = replenishment_report.sort_values(
    by="reorder_quantity_ai",
    ascending=False
).head(10)

labels = top_10["item_id"] + " - " + top_10["store_id"]

plt.figure(figsize=(10, 6))
plt.barh(labels, top_10["reorder_quantity_ai"])

plt.title("Top 10 Replenishment Requirements")
plt.xlabel("Recommended Reorder Quantity")
plt.ylabel("Item - Store")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
store_reorder = replenishment_report.groupby(
    "store_id"
)["reorder_quantity_ai"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
store_reorder.plot(kind="bar")

plt.title("Total Replenishment Requirement by Store")
plt.xlabel("Store")
plt.ylabel("Total Reorder Quantity")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
